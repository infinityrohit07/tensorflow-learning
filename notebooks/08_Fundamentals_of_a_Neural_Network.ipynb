### **What is a neuron?**
A neuron is the basic computational unit of an artificial neural network.

Its job is to receive input values, process them mathematically, and produce an output.

A simple neuron calculates:

z = (x * w) + b

Where:
- \(x\) = Input
- \(w\) = Weight
- \(b\) = Bias
- \(z\) = Output before activation

Let's use a simple example.

Imagine we want to predict a student's exam score based on the number of hours they studied.
- Input: 4 study hours.
- Weight: 3.
- Bias: 2.

```python
# Input
x = 4.0

# Weight
w = 3.0

# Bias
b = 2.0

# Neuron calculation
z = (x * w) + b

print(z)
```

### **Inputs, Weights and Biases**

Inputs:

Inputs are the information we provide to a neural network.

For example, when predicting a student's exam score:
- Study hours
- Hours of sleep
- Previous exam score

These are the input features, usually represented by (x).

Weights (W):

Weights determine how much importance a neuron gives to each input.

For example:
| Input |	Weight |
| --- | --- |
| Study hours	| 0.8 |
| Sleep hours	| 0.2 |
| Previous score | 0.9 |


A larger absolute weight means that the input has a stronger influence on the neuron's calculation, although its actual effect also depends on the input value and other weights.

Bias (b):

Bias is an additional trainable parameter added to the weighted sum of inputs.

It allows a neuron to adjust its output independently of the input values.

**Formula:**

$$
z = (x_1w_1 + x_2w_2 + \cdots + x_nw_n) + b
$$

Where:
- $x$ = Input values
- $w$ = Corresponding weights
- $b$ = Bias
- $z$ = Weighted sum before activation

**Example:**

If the weighted sum is 10 and the bias is 2:

$$
z = 10 + 2 = 12
$$


**Key takeaway:** Weights control the influence of inputs, while bias provides an additional adjustable value that shifts the neuron's output.

### **Layers in a Neural Network**

A neural network is made up of layers. Each layer contains one or more neurons.

There are three main types:
1. Input layer — receives the input features.
2. Hidden layer(s) — process the inputs using weights, biases, and activation functions.
3. Output layer — produces the final prediction.

### **Neural Network Architecture**

Neural network architecture refers to the structure and arrangement of layers and neurons in a neural network.

Imagine our network has:
- Input layer: 3 neurons (study hours, sleep hours, previous score).
- Hidden layer: 4 neurons.
- Output layer: 1 neuron (predicted exam score).

We can represent its architecture as:

$$
3 \rightarrow 4 \rightarrow 1
$$

This means:
- 3 input neurons
- 4 hidden neurons
- 1 output neuron

### **Fully Connected Neural Network**

In a fully connected neural network, every neuron in one layer connects to every neuron in the next layer.

For example, if we have 3 input neurons and 4 hidden neurons, each hidden neuron receives connections from all 3 input neurons.

### **Forward Propagation**

Forward propagation is the process of passing input data through a neural network, layer by layer, to produce a prediction.

The process:
1. Input features enter the input layer.
2. Hidden-layer neurons calculate weighted sums and add biases.
3. Activation functions transform the results.
4. Information moves forward through subsequent layers.
5. The output layer produces the prediction.
