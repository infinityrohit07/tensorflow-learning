# TensorFlow

### What is tensorflow?
**TensorFlow** is an open-source library for **numerical computation** and **machine learning**.

It helps us work with data, perform mathematical operations, and build models that learn patterns from examples.

**Let's understand it with the help of a scenario?**

Imagine if we want to build a program that predicts a value based on an input.

For example, suppose you have these observations:

| Input X | Output y |
|:-------:|:--------:|
| 1       | 5        |
| 2       | 8        |
| 3       | 11       |
| 4       | 14       |
| 5       | 17       |

**Can you see the pattern?**

The output increases by 3 every time the input increases by 1. In fact, these values follow the rule:

**\[y = 3X + 2\]**



We could manually write that formula into a program. But what if we didn't know the rule? What if we had thousands of observations, and the relationship was much more complicated?

That's where our machine learning becomes useful.

Instead of explicitly telling the computer the formula, we give it examples and let a model learn a relationship that predicts the outputs.

### AI vs Machine Learning vs Deep Learning

These three terms are closely related, but they represent different levels of technology.

To understand them properly, let's start from the broadest concept and gradually move towards the most specific one.

The relationship is:

**Artificial Intelligence** → **Machine Learning** → **Deep Learning**

Think of it as three nested concepts:
- AI is the broadest field.
- Machine Learning is a subset of AI.
- Deep Learning is a subset of Machine Learning.

#### **Artificial Intelligence (AI)**

**Artificial Intelligence is the broad field of computer science that focuses on making machines perform tasks that normally require human intelligence.**

Human intelligence involves several abilities:

- Learning from experience.
- Solving problems.
- Making decisions.
- Understanding language.
- Recognizing images and objects.
- Reasoning and planning.

AI attempts to give computers some of these capabilities.

**How does AI work?**

Not every AI system learns from data. Some AI systems work using predefined rules written by programmers. For example, imagine building a simple chatbot.

You could program it with rules:
```python
user_input = "hello"

if user_input == "hello":
    print("Hello! How can I help you?")

elif user_input == "bye":
    print("Goodbye!")
```
Here, the computer isn't learning anything.

It simply follows instructions written by a programmer.

This is an example of a simple rule-based AI approach.

#### **Machine Learning (ML)**

**Machine Learning is a subset of AI that allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed with every individual rule.**

This is where things become particularly interesting.

Instead of manually writing every rule, we provide the computer with examples.

The computer analyzes those examples and learns a mathematical relationship between inputs and outputs.

Let's understand with an example.

Imagine we want to build a model that predicts a student's exam score based on the number of hours they study.

We collect this dataset:
| Study hours (X) | Exam score (y) |
|:---------------:|:--------------:|
| 1               | 5              |
| 2               | 8              |
| 3               | 11             |
| 4               | 14             |
| 5               | 17             |

Here:
- X = Input features (study hours).
- y = Output labels (actual exam scores).
You provide this data to a Machine Learning algorithm.

The algorithm identifies a pattern:

Every additional hour of studying increases the predicted score by 3 points.

It learns this mathematical relationship:
`
[
y = 3x + 2
]`

Now imagine providing a new input:
`
X = 6
`

The trained model predicts:
`[y = 3(6) + 2 = 20]`

Prediction = 20

**Notice something important?**

We never explicitly told the model that the relationship was **y = 3x + 2**.

It discovered that relationship from the training data.

That's the fundamental idea behind Machine Learning.

**How does Machine Learning work?**

A typical Machine Learning workflow looks like this:

1. Collect data
   - Study hours and exam scores


2. Prepare data
   - Clean and organize the dataset


3. Train the model
   - Algorithm learns patterns from data


4. Evaluate the model
   - Check how accurately it predicts results


5. Make predictions
   - Predict scores for new study hours

**Types of Machine Learning**

| Type | Description | Example |
|---|---|---|
| **1. Supervised Learning** | The model learns from labeled data, where both inputs and their correct outputs are provided. | Predicting house prices using house sizes and their corresponding prices. |
| **2. Unsupervised Learning** | The model learns from unlabeled data and discovers hidden patterns or structures. | Grouping customers based on their purchasing behavior. |
| **3. Semi-supervised Learning** | The model learns using a combination of a small amount of labeled data and a large amount of unlabeled data. | Training an image classification model using a few labeled images and many unlabeled images. |
| **4. Reinforcement Learning** | The model learns by interacting with an environment and receiving rewards or penalties for its actions. | Training an AI agent to play a video game by rewarding successful actions. |

#### **Deep Learning (DL)**

**Deep Learning is a specialized subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns from data.**

**What is a Neuron?**

First, let's understand the biological inspiration.

Our brain contains billions of biological cells called neurons. These neurons communicate with each other through electrical and chemical signals. They receive information, process it and pass signals to other neurons.

Artificial neural networks are loosely inspired by this process.

An artificial neuron is a small mathematical unit inside a neural network that receives inputs, performs calculations and produces an output.

**What is a Neural Network?**

A neural network is a collection of interconnected artificial neurons arranged in layers that work together to learn patterns from data.

Instead of relying on one neuron to perform all the calculations, a neural network uses multiple neurons connected together.

**Three Layers of a Neural Network**

A neural network typically consists of three types of layers:

**1. Input Layer**
- The first layer of a neural network.
- Receives input data (features) and passes it to the hidden layer.
- Each input represents a feature from the dataset.

**2. Hidden Layer**
- The layer between the input and output layers.
- Contains neurons that perform mathematical calculations using weights and biases.
- Learns patterns and relationships from input data using activation functions.
- A neural network can have multiple hidden layers.

**3. Output Layer**
- The final layer of a neural network.
- Processes information received from the previous layer.
- Produces the final result, such as a prediction or classification.
- The number of neurons depends on the type of problem being solved.

**Remember:** Input layer receives data, hidden layers learn patterns, and output layer produces the result.

### **Why TensorFlow?**

TensorFlow is an open-source Machine Learning framework that helps us build, train, evaluate and deploy Machine Learning and Deep Learning models.

**Why do we use it?**
- Simplifies the creation of neural networks.
- Handles complex mathematical computations.
- Automatically calculates gradients and updates model weights during training.
- Supports GPU acceleration for faster computations.
- Provides tools to train and deploy models efficiently.

**In simple words:** TensorFlow allows us to build and train neural networks without manually implementing every mathematical operation involved in the learning process.

### **The Five Essential Concepts of Machine Learning**

Before building a Machine Learning model, understand these five concepts:

**1. Data**
- The examples from which a model learns.
- Example: Records containing house sizes and their selling prices.

**2. Features — `X`**
- The input variables used to make predictions.
- Example: House size, location and number of bedrooms.

**3. Labels — `y`**
- The actual answers associated with the training data.
- Example: The actual selling price of each house.

**4. Model**
- A mathematical system that learns relationships between inputs and outputs.
- Example: A neural network that predicts house prices.

**5. Prediction — `ŷ`**
- The output generated by a trained model.
- Written as `ŷ` (y-hat) to distinguish it from the actual value `y`.

<br>

**Connecting the Concepts**

Suppose we want to predict a student's exam score based on study hours.

| Concept | Example |
|---|---|
| `X` (Features) | Hours studied |
| `y` (Label) | Actual exam score |
| Model | Learns the relationship |
| `ŷ` (Prediction) | Predicted exam score |

If the actual score is 80 and the predicted score is 74, the prediction error is 6 points.

<br>

**How Does a Model Learn?**

A Machine Learning model learns through a repeated training process:

**1. Input Data**
- Provide input features (`X`) to the model.

**2. Make Predictions**
- The model generates an output (`ŷ`).

**3. Calculate Loss**
- Measure the difference between predicted and actual values.

**4. Update Parameters**
- Use gradients and an optimizer to adjust weights and biases.

**5. Repeat**
- Continue training over multiple batches and epochs to improve predictions.

<br>

**Important Terms**

| Term | Meaning |
|---|---|
| **Loss** | Measures how incorrect the model's predictions are. |
| **Weights and Biases** | Learnable parameters inside the model. |
| **Optimizer** | Algorithm that updates model parameters using gradients. |
| **Epoch** | One complete pass through the training dataset. |
| **Gradient** | Indicates how changing a parameter affects the loss. |

**Remember:** The basic learning cycle is:

`Predict → Calculate Loss → Update Parameters → Repeat`

We'll explore each of these concepts in detail in later lessons.

### **Our First TensorFlow Code**

````python
# Import TensorFlow and give it the conventional alias "tf".
import tensorflow as tf

# Display the installed TensorFlow version.
print("TensorFlow version:", tf.__version__)

# Create a simple tensor containing three numbers.
numbers = tf.constant([10, 20, 30])

# Display the tensor.
print(numbers)

# Display its shape: one dimension containing three elements.
print("Shape:", numbers.shape)

# Display its data type.
print("Data type:", numbers.dtype)

# Calculate the mean of the three numbers.
print("Mean:", tf.reduce_mean(numbers))
````
