### **Optimizers**

An **optimizer** is an algorithm that updates a neural network's trainable parameters (weights and biases) to minimize the loss during training.

The loss function measures how far the model's predictions are from the actual values, while the optimizer helps the model improve by adjusting its weights and biases.

<br>

#### **How Does an Optimizer Work?**

1. The model makes predictions.
2. The loss function calculates the prediction error.
3. The optimizer uses gradient information to determine how to update the weights and biases.
4. The model's parameters are updated to reduce the loss.
5. This process repeats during training.

<br>

#### **Loss Function vs. Optimizer**

| Loss Function | Optimizer |
|---|---|
| Measures prediction error | Updates model parameters |
| Tells us how wrong the model is | Helps the model improve |
| Examples: MAE, MSE, BCE | Examples: SGD, Adam |

<br>

#### **Common Optimizers**

- **SGD (Stochastic Gradient Descent):** Updates model parameters using gradient information, commonly using one training example or a mini-batch at a time.
- **Adam (Adaptive Moment Estimation):** An optimizer that adapts the learning rate for individual parameters using estimates of gradients and squared gradients.

<br>

#### **Key Points**

- Optimizers help minimize the loss during training.
- They update trainable weights and biases.
- They use gradients to determine parameter updates.
- The learning rate controls the size of parameter updates.
- SGD and Adam are two commonly used optimizers.

### **Gradient Descent**

Gradient Descent is an **optimization algorithm** used to minimize a model's loss by gradually updating its weights and biases.

It uses gradients to determine the direction in which the model's parameters should move to reduce the loss.

<br>

#### **How Does Gradient Descent Work?**

1. The model starts with initial weights and biases.
2. It makes predictions using those parameters.
3. The loss function calculates the prediction error.
4. Gradients are calculated to determine how the loss changes with respect to the parameters.
5. The parameters are updated in the opposite direction of the gradients.
6. This process repeats until the model reaches a suitable minimum.

<br>

#### **Understanding Gradients**

A gradient tells us:
- The direction in which the loss increases.
- How quickly the loss changes when a parameter changes.

Gradient Descent moves in the **opposite direction of the gradient** because its goal is to reduce the loss.

- Negative gradient → Increase the parameter.
- Positive gradient → Decrease the parameter.
- Zero gradient → The slope is flat.

<br>

#### **Gradient Descent Update Rule**
$$
w_{\text{new}}=w_{\text{old}}-\alpha\frac{dL}{dw}
$$

Where:

- $(w_{\text{new}})$ = Updated weight.
- $(w_{\text{old}})$ = Current weight.
- $(\alpha)$ = Learning rate.
- $(\frac{dL}{dw})$ = Gradient of the loss with respect to the weight.

<br>

#### **Example**

Suppose:

- Current weight = 5
- Learning rate = 0.1
- Gradient = 4

Substituting into the formula:
$$
w_{\text{new}}=5-(0.1\times4)
$$

$$
w_{\text{new}}=5-0.4
$$

$$
\boxed{w_{\text{new}}=4.6}
$$

The weight changes from 5 to 4.6, moving in the direction of lower loss.

<br>

### Key Points

- Gradient Descent is used to minimize loss.
- It updates weights and biases using gradients.
- It moves in the opposite direction of the gradient.
- The learning rate controls the size of each update.
- It is a fundamental concept behind optimizers such as SGD and Adam.

### **Learning Rate**

The **learning rate** is a hyperparameter that controls how big a step Gradient Descent takes when updating a model's weights and biases.

It is represented by the Greek letter alpha $(\alpha)$.

<br>

#### **Gradient Descent Update Rule**

$$
w_{\text{new}}=w_{\text{old}}-\alpha\frac{dL}{dw}
$$

Where:

- $(w_{\text{new}})$ = Updated weight.
- $(w_{\text{old}})$ = Current weight.
- $(\alpha)$ = Learning rate.
- $(\frac{dL}{dw})$ = Gradient of the loss with respect to the weight.

<br>

#### **Example: Effect of Different Learning Rates**

Suppose:

- Current weight = 5
- Gradient = 4

Using the update rule:

$$
w_{\text{new}}=5-(\alpha\times4)
$$

| Learning Rate $(\alpha)$ | Calculation | New Weight |
|:---:|:---:|:---:|
| 0.01 | $(5-(0.01\times4))$ | 4.96 |
| 0.1 | $(5-(0.1\times4))$ | 4.6 |
| 0.5 | $(5-(0.5\times4))$ | 3 |
| 1.0 | $(5-(1.0\times4))$ | 1 |

**Observation:** A larger learning rate produces a larger update. However, a larger update is not always better.

<br>

#### **Important Terms**

1. **Minimum:** The point where the loss function reaches its lowest value.
2. **Convergence:** When the optimization process approaches a minimum and the loss stops changing significantly.
3. **Overshooting:** When an update jumps past the minimum instead of moving closer to it.

<br>

#### **Effect of Learning Rate**

| Learning Rate | Effect |
|---|---|
| Very small | Training progresses slowly because parameter updates are small. |
| Appropriate | Helps the model converge efficiently towards a minimum. |
| Very large | May cause the model to overshoot the minimum or diverge. |

<br>

**1. Too Small Learning Rate**
- Takes very small steps.
- Training can become extremely slow.
- May require many iterations to reach a minimum.

**2. Too Large Learning Rate**
- Takes excessively large steps.
- May jump over the minimum.
- Can cause the loss to oscillate or increase instead of decreasing.

**3. Appropriate Learning Rate**
- Takes reasonably sized steps.
- Helps reduce loss steadily.
- Can reach a minimum more efficiently.

<br>

#### **Key Points**

- Learning rate controls the size of parameter updates.
- The gradient determines the direction of the update.
- A very small learning rate can slow down training.
- A very large learning rate can prevent convergence.
- Choosing an appropriate learning rate is important for effective model training.

### **Stochastic Gradient Descent (SGD)**

**Stochastic Gradient Descent (SGD)** is an optimization algorithm that updates a model's weights and biases using the gradient calculated from one training example at a time.

<br>

#### **How SGD Works**

Suppose we have a dataset containing 1,000 training examples.

1. Takes one training example.
2. Makes a prediction.
3. Calculates the loss and gradient.
4. Updates the weights and biases.
5. Repeats with another training example.

So, the model updates its parameters after processing each individual example.

<br>

**Key idea:**

SGD updates model parameters after processing each individual training example instead of waiting for the entire dataset.

### **Types of Gradient Descent**

The main difference between these methods is how many training examples are used to calculate the gradient before updating the model's parameters.

Suppose our dataset contains 1,000 training examples.

**1. Batch Gradient Descent**
- Uses the entire training dataset to calculate the gradient.
- Updates weights and biases once after processing all examples.
- Updates are generally stable but can be computationally expensive for large datasets.

**2. Stochastic Gradient Descent (SGD)**
- Uses one training example to calculate the gradient.
- Updates weights and biases after every example.
- Updates are more frequent but can fluctuate because each example may produce a different gradient.

**3. Mini-batch Gradient Descent**
- Divides the dataset into smaller groups called batches.
- Calculates the gradient using one batch at a time.
- Updates weights and biases after processing each batch.
- Combines the advantages of Batch Gradient Descent and SGD.

<br>

#### **Comparison**

Suppose the dataset contains 1,000 training examples.

| Method | Examples per update | Updates per epoch |
|---|---:|---:|
| Batch Gradient Descent | 1,000 | 1 |
| Stochastic Gradient Descent | 1 | 1,000 |
| Mini-batch Gradient Descent | 32 | Approximately 32 |

<br>

#### **What is an Epoch?**

An **epoch** means one complete pass through the entire training dataset.

For example, if the dataset contains 1,000 training examples, one epoch means the model has processed all 1,000 examples.

#### **Example: Calculating Updates per Epoch**

Suppose:
- Total training examples = 1,000
- Batch size = 100

$$
\text{Updates per epoch}=
\frac{\text{Total examples}}{\text{Batch size}}
=\frac{1000}{100}=10
$$

Therefore, the model performs **10 parameter updates in one epoch**.

<br>

#### **Key Points**

- Batch Gradient Descent uses the entire dataset for each update.
- SGD uses one training example per update.
- Mini-batch Gradient Descent uses a small group of examples per update.
- An epoch represents one complete pass through the dataset.
- Mini-batch training is commonly used in TensorFlow/Keras by specifying `batch_size` in `model.fit()`.

### **Adam Optimizer (Adaptive Moment Estimation)**

**Adam** stands for **Adaptive Moment Estimation**.

It is an optimization algorithm used to update a neural network's weights and biases to reduce the loss during training.

Adam combines ideas from **Momentum** and **RMSprop** to make parameter updates more adaptive and efficient.

<br>

#### **How Does Adam Work?**

Adam tracks two important quantities using past gradients:

**1. First Moment (Momentum)**

- Tracks the exponentially weighted average of past gradients.
- Helps determine the direction in which parameters should move.
- Reduces unnecessary fluctuations in updates.

**2. Second Moment (RMSprop)**

- Tracks the exponentially weighted average of squared past gradients.
- Helps adjust the effective update size for each parameter.
- Parameters with consistently large gradients can receive smaller updates, while those with smaller gradients can receive relatively larger updates.

<br>

**Simple intuition:**

Imagine walking downhill:

- Momentum helps you maintain a consistent direction.
- RMSprop helps adjust your step size.
- Adam combines both ideas to make parameter updates more adaptive.

<br>

#### **Adam vs SGD**

| Feature | SGD | Adam |
|---|---|---|
| Full name | Stochastic Gradient Descent | Adaptive Moment Estimation |
| Gradient history | Basic SGD does not maintain a moving average of past gradients | Tracks past gradients and squared gradients |
| Learning rate | Uses a global learning rate | Adapts effective update sizes for individual parameters |
| Updates | Can fluctuate more | Often produces smoother, more adaptive updates |
| Complexity | Simpler | More complex |


<br>

**Example:**

Suppose a neural network has two weights:

- Weight 1 receives consistently large gradients.
- Weight 2 receives relatively small gradients.

**SGD:**
- Uses the same global learning rate for both weights.
- Their update sizes depend directly on their respective gradients.

**Adam:**
- Adapts the effective update sizes individually using gradient history.
- Can adjust the two weights differently.

<br>

#### **When Should We Use Adam or SGD?**

**SGD:**
- Useful when you want a simpler optimization algorithm.
- Can work well with carefully tuned learning rates and training schedules.
- Sometimes performs better in terms of generalization.

**Adam:**
- Often a good starting choice for neural network training.
- Adapts update sizes for individual parameters.
- Can converge faster in many practical situations.

**Important:** Adam is not always better than SGD. The best optimizer depends on the model, dataset, and training problem.

<br>

### **Learning Rate in Adam**

Adam also uses a learning rate, but it handles parameter updates differently from basic SGD.

- **SGD:** Uses a global learning rate for its parameter updates.
- **Adam:** Uses a global learning rate as a base and adapts the effective update size for each parameter using gradient history.

The commonly used default learning rate in Keras Adam is:

$$
\alpha = 0.001
$$

For comparison, the default learning rate for Keras SGD is:

$$
\alpha = 0.01
$$

These are default values, not universally optimal values. They can be adjusted depending on the model and dataset.

<br>

#### **Implementing Adam in TensorFlow/Keras**

**Creating an Adam optimizer:**

```python
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
```

Explanation:

- `tf.keras.optimizers.Adam()` creates an Adam optimizer.
- `learning_rate=0.001` sets its initial learning rate.
- The optimizer uses gradient information to update trainable parameters.

**Using Adam inside model.compile():**

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mean_squared_error",
    metrics=["mean_absolute_error"]
)
```

Explanation:

- `optimizer`: Determines how weights and biases are updated.
- `loss`: Measures prediction error.
- `metrics`: Track and report model performance during training.

**Using SGD instead:**

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
```

- Adam stands for Adaptive Moment Estimation.
- It combines ideas from Momentum and RMSprop.
- It tracks past gradients and squared gradients.
- It adapts effective update sizes for individual parameters.
- Adam uses a global learning rate as its base learning-rate setting.
- The commonly used default Keras learning rate for Adam is 0.001.
- Adam is often a good starting optimizer for neural networks.
- Adam and SGD are both used to minimize loss by updating trainable parameters.
