### **`tf.random.normal()`**

tf.random.normal() generates a tensor containing random numbers that follow a Normal Distribution (Gaussian Distribution).

A normal distribution is a probability distribution where values tend to cluster around the mean, with fewer values appearing farther away.

For now, just remember that it generates random numbers around a specified average.

```python
# Basic Syntax:

tf.random.normal(
    shape,
    mean=0.0,
    stddev=1.0,
    dtype=tf.float32
)
```

* shape - Shape of the output tensor
* mean - Average value around which numbers are distributed
* stddev - Standard deviation (spread of values)
* dtype - Data type of the output

EXAMPLE 1

````python
A = tf.random.normal(shape=(5,))

print(A)
````

Notice:
- We generated 5 random values.
- The values are generally distributed around 0.
- The output shape is (5,).

EXAMPLE 2

````python
B = tf.random.normal(shape=(2, 3))

print(B)
````

EXAMPLE 3

````python
C = tf.random.normal(
    shape=(5,),
    mean=10,
    stddev=2
)

print(C)
````

Important: Mean is not a guaranteed output value or exact average of the generated samples. It describes the distribution's centre.

Similarly, standard deviation controls the spread, not a strict minimum or maximum.

### **`tf.random.uniform()`**

It generates random numbers where every value within the specified range has an equal probability of being selected.

```python
# Basic Syntax

tf.random.uniform(
    shape,
    minval=0,
    maxval=None,
    dtype=tf.float32
)
```
* shape - Shape of the tensor you want to generate.

* minval - Minimum value of the range.

* maxval - Upper limit of the range (exclusive).

* dtype - Data type of the generated values.

EXAMPLE 1: DEFAULT RANGE

````python
A = tf.random.uniform(shape=(5,))

print(A)
````

EXAMPLE 2: CUSTOM RANGE

````python
B = tf.random.uniform(
    shape=(5,),
    minval=10,
    maxval=20
)

print(B)
````

EXAMPLE 3: DIFFERENT TENSOR SHAPE

````python
C = tf.random.uniform(
    shape=(2,3),
    minval=0,
    maxval=100
)

print(C)
````

### **`tf.random.set_seed()`**

Normally, random functions generate different values each time we run them.

```python
A = tf.random.uniform(shape=(3,))
print(A)
```

Running this multiple times can produce different values.

But sometimes, especially when training machine learning models, we want consistent and reproducible results.

That's where `tf.random.set_seed()` comes in.

Let's see what this produces:

````python
A = tf.random.uniform(shape=(3,))
print(A)
````

Will running it produce the same randomness or other?

````python
A = tf.random.uniform(shape=(3,))
print(A)
````

Ohh, it produces different. Now lets see our `tf.random.set_seed()`

````python
tf.random.set_seed(42)

A = tf.random.uniform(shape=(3,))
print(A)
````

Let's run this again.

````python
tf.random.set_seed(42)

A = tf.random.uniform(shape=(3,))
print(A)
````

You see? It makes the random sequence repeat? Its reproducible.

The seed helps TensorFlow generate a repeatable sequence of random values when the same operations are executed under the same conditions.

Reproducibility helps us:
- Debug models consistently.
- Compare different model configurations fairly.
- Repeat experiments and verify results.

### **Reproducibility**

Reproducibility means being able to run the same experiment again under the same conditions and obtain consistent results.

Imagine we are training a neural network.
1. We train the model and get 92% accuracy.
2. We change something and get 85% accuracy.
3. We want to check whether your change caused the difference.

If random initialization and other random operations vary between runs, comparing results becomes harder.

By controlling randomness with seeds, we can make experiments more consistent.

<br>

**One important distinction:** A seed alone does not guarantee identical results across every device, TensorFlow version, or execution environment. Other sources of nondeterminism may exist.

### **Random Initialization**

Before training a neural network, TensorFlow needs to assign initial values to its weights.

But why random values?

Imagine a neural network with multiple neurons. If every neuron starts with exactly the same weights, they may learn the same things and fail to develop different useful features.

Random initialization gives neurons different starting weights, helping them learn different patterns.

````python
weights = tf.random.normal(shape=(3, 2))

print(weights)
````

This creates a weight matrix with 3 rows and 2 columns, initialized with random values.

**Remember:** Random initialization gives the model a starting point. Training then adjusts those weights to improve predictions.

### **Why do neural networks need random weights?**

Imagine we have three neurons, and all three start with identical weights:

```python
weights = [0.5, 0.5, 0.5]
```

If they receive the same inputs and use the same activation function, they will produce the same outputs and receive the same updates during training.

As a result, they won't learn different features. This is called the **symmetry problem**.

<br>

**How does random initialization help?**

Instead, we start with different weights:
```python
weights = [0.2, -0.7, 0.9]
```

Now the neurons have different starting points and can learn different patterns.

<br>

**Important distinction**

Random initialization does not mean assigning large random values.

If weights are too large or too small, training can become difficult. That's why neural networks use carefully designed initialization methods, such as:
- Glorot/Xavier initialization
- He initialization

We'll explore these later when we study neural network layers and training in more depth.

### **Numerical Precision**

Numerical precision refers to how accurately a computer can represent and perform calculations with numbers.
- Computers use floating-point representation to store decimal numbers.
- Some decimal numbers cannot be represented exactly in binary, resulting in small differences called floating-point errors.
- float32 provides approximately 6–7 significant decimal digits of precision.
- float64 provides approximately 15–16 significant decimal digits of precision.
- float64 offers greater precision but uses more memory.
- Instead of comparing floating-point values using exact equality (==), we can use a tolerance-based comparison to check whether two values are sufficiently close.

````python
a = tf.constant(0.1, dtype=tf.float32)
b = tf.constant(0.2, dtype=tf.float32)

result = a + b

print(result)
print(result.numpy() == 0.3)
````

````python
print(tf.cast(result, tf.float64).numpy())
print(result.numpy == tf.cast(result, tf.float64).numpy())
````
