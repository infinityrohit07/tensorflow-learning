### **What is a tensor?**

A tensor is a data structure that stores numerical values in one or more dimensions.

Before we move forward, let's import our tensorflow so we can code and understand our topics.

```python
import tensorflow as tf
print(tf.__version__)
```

### `tf.constant()`

Since we are gonna start creating our tensors using `tf.constant()`, it's important for us to know what `tf.constant()` is.

`tf.constant()` is the core function used to create immutable tensors in TensorFlow. It converts Python lists, tuples, or NumPy arrays into a fixed multidimensional array that cannot be altered or updated in place.

<br>

### **Types of Tensors**

**Scalar - 0 Dimension**
- A single number. No rows, no columns, no axes.

```python
scalar = tf.constant(25)
print("Scalar tensor: ", scalar)
```

```python
print(scalar.numpy())
```

<br>

**Vector — 1 Dimension**
- A sequence of numbers arranged along one axis.

```python
vector = tf.constant([10, 20, 30])
print("Vector tensor:", vector)
```

```python
print(vector.numpy())
```

<br>

**Matrix — 2 Dimensions**
- Rows and columns: two axes.

```python
matrix = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])
print("Matrix:", matrix)
```

```python
matrix.numpy()
```

<br>

**3D tensor — 3 Dimensions**
- Multiple matrices arranged along another axis.

```python
tensor_3d = tf.constant([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("3D tensor:", tensor_3d)
```

```python
tensor_3d.numpy()
```

<br>

**Higher-dimensional tensors**
- Tensors with more than three dimensions are often used to represent more complex data, such as color images.

<br>

### **What is a rank?**

Rank tells us how many dimensions (axes) a tensor has.

```python
# tf.rank() returns the number of dimensions.
print("Scalar rank:", tf.rank(scalar))
print("Vector rank:", tf.rank(vector))
print("Matrix rank:", tf.rank(matrix))
print("3D Tensor rank:", tf.rank(tensor_3d))
```

In the output, before the shape, 0 or 1 or 2 or 3 are representing the ranks of each tensor.

If we want to see it more clearly we can convert it to numpy like:

```python
# tf.rank() returns the number of dimensions.
print("Scalar rank:", tf.rank(scalar).numpy())
print("Vector rank:", tf.rank(vector).numpy())
print("Matrix rank:", tf.rank(matrix).numpy())
print("3D Tensor rank:", tf.rank(tensor_3d).numpy())
```

<br>

### **What is a Shape?**
Shape tells us how many elements exist along each dimension of a tensor.

```python
print("Shape of scalar:", scalar.shape)
print("Shape of vector:", vector.shape)
print("Shape of matrix:", matrix.shape)
print("Shape of 3D Tensor:", tensor_3d.shape)
```

Understanding the Output

The output shows the **shape of each tensor**, which tells us how many elements exist along each dimension.

- **Scalar `()`**: A scalar has 0 dimensions because it contains only a single value. Therefore, its shape is an empty tuple `()`.

- **Vector `(3,)`**: A vector has 1 dimension and contains 3 elements. The comma indicates that it is a one-dimensional tuple.

- **Matrix `(2, 3)`**: A matrix has 2 dimensions, containing 2 rows and 3 columns.

- **3D Tensor `(2, 2, 2)`**: A 3D tensor has 3 dimensions. It contains 2 matrices, each having 2 rows and 2 columns.

**Key Takeaways:**
- The number of values in the shape tuple represents the number of dimensions.
- Each value represents the number of elements along that dimension.
- To calculate the total number of elements, multiply all the values in the shape.

For example, a tensor with shape `(2, 2, 2)` contains:

`2 × 2 × 2 = 8 elements`

### **What is an Axis in TensorFlow?**
An axis refers to a specific dimension of a tensor. It tells us along which dimension we want to perform an operation.

```python
print(matrix.numpy())
print("Shape of matrix:", matrix.shape)
```

Its shape is (2, 3):
- Axis 0 represents the rows (2 rows).
- Axis 1 represents the columns (3 columns).

Now performing sum along Axis 0 and Axis 1.
Note: we have used `tf.reduce_sum()` for finding sum along an axis.

```python
# the matrix
print(matrix.numpy())

print("---------------------------------------")

# Sum along axis 0
print(tf.reduce_sum(matrix, axis=0).numpy())

print("---------------------------------------")

# Sum along axis 1
print(tf.reduce_sum(matrix, axis=1).numpy())
```

- Axis 0 - we have added elements vertically (column-wise).

- Axis 1 - we have add elements horizontally (row-wise).
