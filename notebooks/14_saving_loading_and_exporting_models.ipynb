### **We trained a model**

````python
import tensorflow as tf

X = tf.constant([1, 2, 3, 4, 5], dtype=tf.float32)

y = tf.constant([20, 30, 40, 50, 60], dtype=tf.float32)

X,y
````

````python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mean_squared_error",
    metrics=["mean_absolute_error"]
)
````

````python
history = model.fit(X, y, epochs=200)
````

````python
X_new = tf.constant([6, 7, 8], dtype=tf.float32)

predictions = model.predict(X_new)

print(predictions)
````

### **Saving, Loading and Exporting Models**

**Why Do We Need to Save Trained Models?**

During training, a neural network learns values for its weights and biases.

These learned parameters are stored in the model's memory.

However, when a Google Colab runtime disconnects or resets, the variables and objects stored in memory are lost.

Saving a model allows us to:

- Reuse a trained model without training it again.

- Avoid wasting time and computational resources.

- Share trained models with others.

- Use trained models in applications and deployment environments.

- Continue working with a model in a future session.

**Important:** A saved model allows us to restore its learned parameters and use it without repeating the original training process.

#### **Saving Models Using `model.save()`**

TensorFlow Keras provides the `model.save()` method to save a complete trained model.

Syntax:

`model.save("filename.keras")`

````python
model.save("my_first_model.keras")
````

Here:

- model is our trained model.

- .save() saves the complete model.

- "my_first_model.keras" is the filename.

##### **Checking Whether the File Exists**

````python
import os

os.path.exists("my_first_model.keras")
````

True confirms that the model file exists.

**Where Is the File Stored in Google Colab?**

By default, the file is saved in the current working directory, usually:

`/content/my_first_model.keras`

**Important:** Colab's temporary runtime storage is not permanent. Files stored there can disappear when the runtime is deleted.

**Understanding the .keras File Format**

The .keras format is the recommended native format for saving complete Keras models.

Internally, a .keras file is a ZIP archive containing files such as:
```
my_first_model.keras
│
├── metadata.json
├── config.json
└── model.weights.h5
```

**Purpose of Each File**

|File | Purpose |
|:---:|:---:|
|metadata.json|Stores metadata, including the Keras version.|
|config.json|Stores the model architecture and configuration.|
|model.weights.h5|Stores learned weights, biases and other saved variable values.|

##### **Inspecting the .keras Archive:**

````python
import zipfile

with zipfile.ZipFile("my_first_model.keras", "r") as file:
    print(file.namelist())
````

Together, these three allow Keras to reconstruct the saved model.

#### **Loading Saved Models**

We don't need to train a model again just to use it.

We can restore a previously saved model using `tf.keras.models.load_model()`.

Syntax:

`loaded_model = tf.keras.models.load_model("filename.keras")`

````python
loaded_model = tf.keras.models.load_model("my_first_model.keras")
````

Here:

- load_model() loads the saved model.

- "my_first_model.keras" is the saved model file.

- loaded_model is the variable that holds the restored model.

##### **Verifying the Loaded Model:**

````python
loaded_model.summary()
````

Our Dense layer uses the formula:

$$
y = wx+b
$$

Therefore, it has:

- 1 weight (w)

- 1 bias (b)

<br>

Total trainable parameters = 2.

The additional 6 parameters belong to the saved Adam optimizer state. They are not additional Dense-layer weights.

<br>

**Important:** The loaded model restores the learned parameters, so we can use it without training again.

#### **Making Predictions Using a Loaded Model**

After loading a model, we can make predictions using its `predict()` method.

````python
predictions = loaded_model.predict(tf.constant([6, 7, 8], dtype=tf.float32))

print(predictions)
````

### **What Did We Achieve?**

1. Trained a model.

2. Saved the complete model using `model.save()`.

3. Inspected the .keras archive.

4. Loaded the saved model using `load_model()`.

5. Made predictions using the restored model without calling `model.fit()` again.

### **Saving and Loading Model Weights Separately**

So far, we have saved the complete model using model.`save()`.

However, sometimes we only want to save the learned weights and biases instead of saving the entire model.

For this, Keras provides:

`save_weights()` — saves only the model's learned weights.

`load_weights()` — restores previously saved weights into a model.

#### **Saving Only the Weights**

````python
model.save_weights("my_model.weights.h5")
````

Here:

- model is our trained model.

- `save_weights()` saves its learned weights and biases.

- .weights.h5 is the required filename convention in Keras 3.

 #### **Checking Whether the File Exists**

````python
os.path.exists("my_model.weights.h5")
````

True confirms that the weights file exists.

#### **Loading Saved Weights**

Unlike loading a complete model, loading weights requires us to recreate the model architecture first.

This is because the weights file does not contain the complete model architecture.

##### **1. Create a new model with the same architecture**

````python
new_model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
````

This creates a new model with:

- One input feature.

- One Dense layer.

- One output neuron.

Its weights are initially newly initialized.

##### **2. Load the saved weights**

````python
new_model.load_weights("my_model.weights.h5")
````

This restores the learned weights and biases from the saved file into new_model.

##### **3. Make predictions**

````python
predictions = new_model.predict(
    tf.constant([6, 7, 8], dtype=tf.float32)
)

print(predictions)
````

### **What Is Model Portability?**

Model portability means being able to take a trained model from one environment and use it in another environment.

For example:

1. Train a model in Google Colab.

2. Save it as my_first_model.keras.

3. Download the file or store it in Google Drive.

4. Load it on another computer.

5. Use it to make predictions without training it again.

<br>

### **Why Is Model Portability Useful?**

Imagine we train a machine learning model in Google Colab but want to use it in a web application.

We don't need to train the model every time someone visits our website.

Instead, we can:

- Train the model once.

- Save the trained model.

- Load it in the application environment.

- Use it to generate predictions.

This saves time and computational resources.
