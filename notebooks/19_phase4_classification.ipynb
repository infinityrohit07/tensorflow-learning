# **Evaluation of Classification Models**

## **1. What is Accuracy?**

**Accuracy** measures the proportion of total predictions that the model classified correctly.

It tells us how often the model makes correct predictions overall.

#### **Formula :**

$$
Accuracy = \frac{Correct\ Predictions}{Total\ Predictions}
$$

#### **Example :**

Suppose a model makes 100 predictions:
- 80 correct predictions
- 20 incorrect predictions

$$
Accuracy = \frac{80}{100}=80\%
$$

#### **Limitation of Accuracy**

Accuracy can be misleading when the dataset is imbalanced.

Example:

Suppose a dataset contains:
- 95 students who passed.
- 5 students who failed.

If the model predicts Pass for everyone, its accuracy is 95%, but it fails to identify anyone who actually failed.


<br>

## **2. What is a Confusion Matrix?**

A **Confusion Matrix** is a table used to evaluate a classification model by comparing actual labels with predicted labels.

It helps us understand what the model predicted correctly and where it made mistakes.

<br>

#### **Four Components :**

For binary classification:
- Positive (1): Pass
- Negative (0): Fail

<br>

| Component | Meaning |
|---|---|
| True Positive (TP) | Actual 1, Predicted 1 |
| True Negative (TN) | Actual 0, Predicted 0 |
| False Positive (FP) | Actual 0, Predicted 1 |
| False Negative (FN) | Actual 1, Predicted 0 |

**Remember:**
- True / False tells us whether the prediction was correct.
- Positive / Negative tells us which class was predicted.

<br>

### **Confusion Matrix Structure**

Rows represent actual labels, and columns represent predicted labels.

| | Predicted Positive (1) | Predicted Negative (0) |
|---|---:|---:|
| Actual Positive (1) | TP | FN |
| Actual Negative (0) | FP | TN |

<br>

### **Creating a Confusion Matrix using Scikit-learn**

```python
from sklearn.metrics import confusion_matrix

y_true = [1, 0, 1, 1, 0, 0, 1, 0]
y_pred = [1, 0, 0, 1, 1, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

print(cm)
```

**Output:**

```text
[[3 1]
 [1 3]]
```

Scikit-learn uses this arrangement by default:
- Rows = Actual labels (0, 1)
- Columns = Predicted labels (0, 1)


| | Predicted 0 | Predicted 1 |
|---|---:|---:|
| Actual 0 | TN = 3 | FP = 1 |
| Actual 1 | FN = 1 | TP = 3 |

<br>

### **Visualizing the Confusion Matrix**

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")
plt.show()
```

- `annot=True` displays values inside the cells.
- `fmt="d"` displays values as integers.
- `cmap="Blues"` sets the heatmap color scheme.

<br>

#### **Calculations using a Confusion Matrix**

**Total Predictions**

$$
Total = TP+TN+FP+FN
$$

**Correct Predictions**

$$
Correct = TP+TN
$$

**Incorrect Predictions**

$$
Incorrect = FP+FN
$$

**Accuracy**

$$
Accuracy = \frac{TP+TN}{TP+TN+FP+FN}
$$

Example:

If TP = 40, TN = 30, FP = 20 and FN = 10:

- Total predictions = 100
- Correct predictions = 70
- Incorrect predictions = 30
- Accuracy = 70%

<br>

## **3. What is Precision?**

**Precision** tells us how many of the positive predictions made by the model were actually positive.

In simple words:

Out of everything the model predicted as positive, how much was correct?

#### **Formula :**

$$
Precision = \frac{TP}{TP+FP}
$$

Where:
- TP = True Positives
- FP = False Positives

#### **Example :**

Suppose a model predicts 20 emails as Spam:
- 15 are actually Spam.
- 5 are Not Spam.

$$
Precision = \frac{15}{15+5}
$$

$$
=\frac{15}{20}=\boxed{75\%}
$$

This means 75% of the emails predicted as Spam were actually Spam.

#### **Why is Precision important?**

Precision is useful when we want to avoid incorrectly predicting negative cases as positive.

Examples:
- Spam detection: Avoid marking legitimate emails as Spam.
- Fraud detection: Avoid incorrectly flagging normal transactions as fraudulent.

**Key Takeaway:**

High Precision means fewer False Positives (FP).

<br>

## **4. What is Recall?**

**Recall** tells us how many of the actual positive cases the model successfully identified.

In simple words:

Out of all the actual positive cases, how many did the model find?

#### **Formula :**

$$
Recall = \frac{TP}{TP+FN}
$$

Where:
- TP = True Positives
- FN = False Negatives

#### **Example :**

Suppose there are 100 actual positive cases:
- 80 are correctly identified.
- 20 are missed by the model.

$$
Recall = \frac{80}{80+20}
$$

$$
=\frac{80}{100}=\boxed{80\%}
$$

This means the model successfully identified 80% of all actual positive cases.

#### **Why is Recall important?**

Recall is useful when missing actual positive cases can have serious consequences.

Examples:
- Medical diagnosis: Identify as many patients with a disease as possible.
- Fraud detection: Catch as many fraudulent transactions as possible.
- Spam detection: Find as many actual spam emails as possible.

**Key Takeaway:**

High Recall means fewer False Negatives (FN).

#### **Precision vs Recall**

| Metric | Formula | Focus |
|---|---|---|
| Precision | TP / (TP + FP) | Reliability of positive predictions |
| Recall | TP / (TP + FN) | Finding actual positive cases |

<br>

## **5. What is F1 Score?**

**F1 Score** combines Precision and Recall into a single evaluation metric.

It helps us understand how well a model balances both metrics.

#### **Why do we need F1 Score?**

Sometimes:
- A model has high Precision but low Recall.
- A model has high Recall but low Precision.

Looking at only one metric can hide an important weakness.

F1 Score helps us evaluate both together.

#### **Formula :**

F1 Score is the harmonic mean of Precision and Recall.

$$
F1 = 2\times\frac{Precision\times Recall}{Precision+Recall}
$$

#### **Example :**

Suppose:
- Precision = 80%
- Recall = 60%

$$
F1 = 2\times\frac{0.8\times0.6}{0.8+0.6}
$$

$$
=\boxed{0.686=68.6\%}
$$

#### **Understanding F1 Score**

Consider a fraud detection model:

- TP = 90
- FP = 10
- FN = 90

Precision:

$$
\frac{90}{90+10}=90\%
$$

Recall:

$$
\frac{90}{90+90}=50\%
$$

F1 Score:

$$
2\times\frac{0.9\times0.5}{0.9+0.5}
=\boxed{64.29\%}
$$

Although Precision is 90%, Recall is only 50%. F1 Score reflects the imbalance between them.

#### **When do we need F1 Score?**

F1 Score is particularly useful when both False Positives and False Negatives matter.

Examples:
- Fraud detection
- Spam detection
- Medical diagnosis

**Key Takeaways**
- F1 Score combines Precision and Recall.
- Higher F1 generally indicates a better balance between them.
- It does not directly include True Negatives.
- F1 treats Precision and Recall equally.

## **6. What is an ROC Curve?**

ROC stands for **Receiver Operating Characteristic**.

An ROC Curve is a graph that shows how well a binary classification model distinguishes between positive and negative classes at different classification thresholds.

It plots:

- X-axis: False Positive Rate (FPR)
- Y-axis: True Positive Rate (TPR)

#### **True Positive Rate (TPR)**

TPR is also known as Recall.

It measures how many actual positive cases the model correctly identifies.

$$
TPR = \frac{TP}{TP+FN}
$$

#### **False Positive Rate (FPR)**

FPR measures how many actual negative cases the model incorrectly classifies as positive.

$$
FPR = \frac{FP}{FP+TN}
$$

#### **How does the Classification Threshold affect the ROC Curve?**

Remember that the classification threshold converts predicted probabilities into class labels.

For example:

- Probability >= threshold → Class 1
- Probability < threshold → Class 0

Consider these predicted probabilities:

```python
y_true = [1, 1, 0, 0, 1, 0]

y_pred = [0.90, 0.70, 0.60, 0.40, 0.30, 0.10]
```

At threshold 0.5:

| Actual | Probability | Predicted |
|---:|---:|---:|
| 1 | 0.90 | 1 |
| 1 | 0.70 | 1 |
| 0 | 0.60 | 1 |
| 0 | 0.40 | 0 |
| 1 | 0.30 | 0 |
| 0 | 0.10 | 0 |

The confusion matrix components are:
- TP = 2
- FN = 1
- FP = 1
- TN = 2

Therefore:

$$
TPR = \frac{2}{2+1}=0.667
$$

$$
FPR = \frac{1}{1+2}=0.333
$$

Our ROC point is:

$$
(FPR,TPR)=(0.333,0.667)
$$

#### **Effect of changing the Threshold**

| Threshold | General effect |
|---|---|
| Lower threshold | More positive predictions |
| Higher threshold | Fewer positive predictions |

Generally:
- Lower threshold increases TPR but can also increase FPR.
- Higher threshold decreases FPR but can also decrease TPR.

The ROC Curve plots the TPR and FPR values obtained at different thresholds.

#### **Understanding the ROC Curve**

The ideal ROC point is near the **top-left corner**.

Why?

- TPR should be close to 1 (identify more actual positives).
- FPR should be close to 0 (make fewer false positive predictions).

## **7. What is AUC?**

AUC stands for **Area Under the Curve**.

It measures how well a classification model can distinguish between positive and negative classes.

| AUC | Interpretation |
|---:|---|
| 1.0 | Perfect discrimination |
| 0.9 | Excellent discrimination |
| 0.8 | Good discrimination |
| 0.7 | Fair discrimination |
| 0.5 | Random guessing |
| Below 0.5 | Worse than random ranking |

These are common rules of thumb, not universal grading standards.

**Important:** An AUC of 0.95 does not mean the model has 95% accuracy. It indicates excellent ability to distinguish between positive and negative classes.

<br>

#### **Key Takeaways**

- ROC Curve plots TPR against FPR.
- TPR is the same as Recall.
- FPR measures the proportion of actual negatives incorrectly classified as positive.
- Changing the classification threshold changes TPR and FPR.
- AUC measures the model's ability to distinguish between classes.
- Higher AUC generally indicates better class discrimination.
