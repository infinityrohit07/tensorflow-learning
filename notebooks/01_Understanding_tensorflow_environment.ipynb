### **Importing Tensorflow and checking its environment**

```python
# Import TensorFlow.
import tensorflow as tf

# Display the TensorFlow version available in this runtime.
print("TensorFlow version:", tf.__version__)

# Check whether TensorFlow can access a CPU device.
print("CPU devices:", tf.config.list_physical_devices("CPU"))

# Check whether TensorFlow can access a GPU device.
print("GPU devices:", tf.config.list_physical_devices("GPU"))
```

`import tensorflow as tf`
- Imports TensorFlow and gives it the shorter name tf. This is the standard convention used in TensorFlow tutorials and projects.

`tf.__version__`
- Returns the installed TensorFlow version. For example, it might display 2.x.x; the exact version depends on the current Colab runtime.

`tf.config.list_physical_devices("CPU")`
- Lists CPU devices available to TensorFlow.

`tf.config.list_physical_devices("GPU")`
- Lists GPU devices available to TensorFlow.

### **Understanding CPU and GPU in Google Colab**

Google Colab provides different hardware options for running your code.

**1. CPU (Central Processing Unit)**
- General-purpose processor.
- Suitable for basic tensor operations and smaller Machine Learning models.
- Default hardware option in many Colab sessions.

**2. GPU (Graphics Processing Unit)**
- Designed to perform many calculations simultaneously.
- Can accelerate tensor operations and neural network training.
- Particularly useful for larger datasets and Deep Learning models.
- Availability depends on your Colab account, usage limits and available resources.

<br>

**How to Enable a GPU in Google Colab**

1. Open **Runtime** in Colab.
2. Select **Change runtime type**.
3. Under **Hardware accelerator**, choose **GPU** (if available).
4. Click **Save**.
5. Run your device-checking code again.

<br>

**Remember:**
- CPU is sufficient for learning TensorFlow basics and running small models.
- GPU can significantly speed up computationally intensive Deep Learning tasks.
- You don't need a GPU for every TensorFlow project.
