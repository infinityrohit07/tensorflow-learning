### **What does creating a tensor mean?**
A tensor is a container for numerical or other data, organized into dimensions.

For example:
- Scalar: a single number.
- Vector: a one-dimensional collection.
- Matrix: a two-dimensional collection.
- Higher-dimensional tensor: data arranged across three or more dimensions.
TensorFlow provides several functions to create these tensors without manually writing every value.

#### `tf.constant()`
- Creates a tensor with specific values
- we have seen this function before, right?


````python
# Creating a tensor containing specific values
tensor = tf.constant([10, 20, 30, 40])

print(tensor)
````

it created a one dimension tensor with 5 element.

<br>

#### `tf.zeros()`

- It creates a tensor where every element starts at zero.

- tf.zeros(shape, dtype=tf.float32)

````python
# Creating a vector containing 5 zeros
tensor = tf.zeros(shape=(5,))
print(tensor)
````

It created a tensor of one dimension with 5 elements.

<br>

#### `tf.ones()`

- it creates a tensor where every elements starts at one.
- tf.ones(shape, dtype=tf.float32)


````python
# Creating a vector containing 5 ones.
tensor = tf.ones(shape=(5,))
print(tensor)
````

````python
# Creating a matrix with 3 rows and 2 columns
tensor = tf.ones(shape=(3, 2))
print(tensor)
````

<br>

#### `tf.fill()`
- Creates a tensor filled with a specific value
- tf.fill(dims, value)

````python
# Creating a matrix filled with 5
tensor = tf.fill(
    dims=(2, 3),
    value=5
)

print(tensor)
````

<br>

#### `tf.range()`
- Creates a sequence of numbers
- tf.range(start, limit, delta)

```
start -	Starting number (inclusive)
limit -	Ending boundary (exclusive)
delta	- Difference between consecutive numbers

````python
tensor = tf.range(10)
print(tensor)
````

it simply created a 1D tensor with 10 elements

````python
tensor = tf.range(start=2, limit=10)
print(tensor)
````

it created a 1D tensor starting from 2 and ending at 9, one before 10.

````python
tensor = tf.range(start=0, limit=21, delta=5)
print(tensor)
````

it created a 1D tensor staring from 0 and ending at 21 as limit says but with 5 difference between each consecutive numbers

<br>

#### `tf.zeros_like() and tf.ones_like()`

- These functions create a new tensor based on the shape of an existing tensor.

````python
# Original tensor
original = tf.constant([[1, 2, 3],
                         [4, 5, 6]])

# Creating another tensor with the same shape, filled with zeros
zeros = tf.zeros_like(original)

print(zeros)
````

````python
# Creating a tensor with the same shape, filled with ones
ones = tf.ones_like(original)

print(ones)
````

### **Creating tensor with** `tf.Variable()`

`tf.constant()` is immutable, we cannot directly change its values. But whar if we want to create a tensor whose values can be changed?

In machine learning, models need to update their values while learning from data which means we need a tensor whose value can be changed.

This is where `tf.Variable()` comes into use. It creates a mutable tensor whose value can be updated after creation.

<br>

Basic Syntax:
```python
tf.Variable(initial_value, dtype=None, name=None)
```


SCALAR

````python
weight = tf.Variable(10)

print(weight)
````

VECTOR

````python
numbers = tf.Variable([10, 20, 30, 40])

print(numbers)
````

MATRIX

````python
matrix = tf.Variable([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
````

we can create scalars, vectors, and matrices using `tf.Variable()`, just like `tf.constant()`.

### `tf.constant() vs tf.Variable()`

The difference is that variables can be updated.

````python
# Creating a constant
constant_tensor = tf.constant([10, 20, 30])

# Creating a variable
variable_tensor = tf.Variable([10, 20, 30])
````

Let's try to update our constant trensor first.

````python
# This will raise an error
constant_tensor.assign([40, 50, 60])
````

Why does this fail?

Because a regular tf.constant() tensor does not have an assign() method. Its values cannot be modified in place.

Now, let's update the variable tensor

````python
variable_tensor.assign([40, 50, 60])

print(variable_tensor)
````

This works because `tf.Variable()` supports updating its values.
