### **Converting TensorFlow tensors into Numpy arrays**

Before we begin, remember:

- TensorFlow uses tensors to store and process data.
- NumPy uses ndarrays to store and process numerical data.

Both can represent multidimensional numerical data, and we can convert between them.

````python
import numpy as np

A = tf.constant([10, 20, 30, 40])
print(A)
print(type(A))

print()

# Converting the tensor into a NumPy array
B = A.numpy()
print(B)
print(type(B))
````

<br>

What does `.numpy()` do?

The `.numpy()` method converts a TensorFlow tensor into a NumPy array.

<br>

Before Conversion:
```
<class 'tensorflow.python.framework.ops.EagerTensor'>
```
This tells us that A is a TensorFlow tensor (EagerTensor).

<br>

After Conversion:
```
<class 'numpy.ndarray'>
```
Now B is a NumPy ndarray, not a TensorFlow tensor.

### **Converting NumPy arrays into TensorFlow tensors**

We can convert a NumPy array into a TensorFlow tensor using `tf.constant()`.

````python
A = np.array([10, 20, 30, 40])
print(A)
print(type(A))

print()

# Converting NumPy array into a TensorFlow tensor
B = tf.constant(A)
print(B)
print(type(B))
````

You can convert a NumPy array into both a TensorFlow constant and a TensorFlow variable.

````python
C = tf.Variable(A)
print(C)
print(type(C))
````

### **Eager Execution in TensorFlow**

**Eager execution** means TensorFlow executes operations immediately, as soon as you run them, instead of building a computation graph first.

In simple words, you write an operation, execute the cell, and immediately see its result.

TensorFlow uses eager execution by default.

````python
A = tf.constant([10, 20, 30])
B = tf.constant([1, 2, 3])

# Performing addition
C = A + B

# Displaying the result
print(C)
````

Running above code immediately performed the addition and gives us the result.

That's eager execution.

#### **How do we know eager execution is enabled??**

TensorFlow provides a function:

````python
print(tf.executing_eagerly())
````

#### **Why is eager execution useful?**
- Immediate results: You can see outputs immediately.
- Easy debugging: You can inspect tensors and variables while executing code.
- Beginner-friendly: It behaves similarly to regular Python execution.
- Interactive development: You can experiment with TensorFlow operations easily in Colab.

### NumPy and TensorFlow Operations Together

TensorFlow tensors and NumPy arrays can work together in mathematical operations.

````python
# Creating a TensorFlow tensor
A = tf.constant([10, 20, 30])

# Creating a NumPy array
B = np.array([1, 2, 3])

# Performing addition
C = A + B

print(C)
print(type(C))
````

Important: The result is a TensorFlow tensor, not a NumPy array.

````python
A = tf.constant([20, 30, 40])
B = np.array([2, 3, 4])

C = A - B

print(C)
print(type(C))
````
